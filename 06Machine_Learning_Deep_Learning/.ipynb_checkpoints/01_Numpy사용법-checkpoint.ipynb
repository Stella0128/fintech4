# numpy
* [https://numpy.org/devdocs/user/quickstart.html](https://numpy.org/devdocs/user/quickstart.html)
* numpy: Numerical Python 과학적, 산업적 연산을 위한 패키지
* numpy는 python의 list와 비슷한 자료형

# 차원
* 0 차원 = 점 . 스칼라  a = 10
* 1 차원 = 선 --- 방향이 없는 벡터, 배열, list, tuple  [1, 2, 3, 4]
* 2 차원 = 면(가로, 세로) - DB 테이블, DataFrame, excel Sheet, matrix, 행렬
* 3 차원 = 입체(가로, 세로, 깊이), 정육면체(주사위), 2차원 데이터의 묶음 - excel file
* 4 차원 = 3차원 데이터의 묶음
* 5 차원 = 4차원 데이터의 묶음

# Numpy를 사용해야 하는 이유
* 행렬 연산이 쉽고 빠르기 때문

```python
# 0차원 스칼라 연산
a = 10
b = 5
scalar = a + b
scalar
```

```python
scalar + 10
```

```python
# 1차원 벡터의 연산
v1 = [1,2,3,4,5]
# v1의 요소에 10을 곱한 리스트를 만드세요.
v1 * 10
```

```python
# v1의 요소에 10을 곱한 리스트를 만드세요.
mul10 = []
for num in v1:
    mul10.append(num * 10)
mul10
```

```python
mul10 = [num * 10 for num in v1]
mul10
```

```python
# 2차원 데이터 연산 10 곱해서 출력
d2
```

```python
for l1 in d2:
    print(l1)
```

```python
for l1 in d2:
    for num in l1:
        print(num)
```

```python
for l1 in d2:
    for num in l1:
        print(num * 10)
```

```python
result = []
for l1 in d2:
    temp = []
    for num in l1:
        temp.append(num * 10)
    result.append(temp)
result
```

```python
# 3차원 데이터에 10 곱하는 연산
```

```python
for item1 in d3:
    print(item1)
```

```python
for item1 in d3:
    for item2 in item1:
        print(item2)
```

```python
for item1 in d3:
    for item2 in item1:
        for item3 in item2:
            print(item3)
```

```python
result3 = []
for item1 in d3:
    temp2 = []
    for item2 in item1:
        temp1 = []
        for item3 in item2:
            temp1.append(item3 * 10)
        temp2.append(temp1)
    result3.append(temp2)
result3 = np.array(result3)
print(result3, "\n", result3.ndim, result3.shape)
```

```python
d4 * 10
```

# numpy에서 자주 사용하는 메서드
* .ndim : 몇 차원인지 확인
* .shape : 몇 차원, 몇행, 몇열인지 구조 확인
* .size : 배열 안에 있는 요소의 총 개수가 몇 개인지
* .dtype : 데이터 타입이 무엇인지

```python
d5
```

```python
print(d5.ndim, d5.shape, d5.size, d5.dtype)
```

# ndarray 만들기
* numpy의 기본 자료형 ndarray
* np.array(리스트/튜플)
* 자료형은 반드시 1가지만 넣는다.

```python
e1 = np.array([1,2,3,4,5])
e1
```

```python
e2 = np.array((1,2,3,4,5))
e2
```

```python
e3 = np.array((1,2,3.0,4,5))
print(e3, e3.dtype)
```

```python
e4 = np.array((1,2,3.0,"4",5))
print(e4, e4.dtype)
```

# np.arange(시작, 끝, step)

```python
a = np.arange(1, 11, 2)
a
```

```python
a2 = np.arange(11, 1, -3)
a2
```

```python
# 1-100까지 있는 array
a3 = np.arange(1, 101)
a3
```

# .reshape(행, 열)
* 벡터를 원하는 크기의 다차원 배열로 변환

```python
print(a3.ndim, a3.shape)
```

```python
a4 = a3.reshape(20, 5)
print(a4.ndim, a4.shape)
a4
```

```python
a4 = a3.reshape(5, 2, 2, 5)
print(a4.ndim, a4.shape)
a4
```

```python
a5 = a3.reshape(-1, 1, 2, 2, 5)
print(a5.ndim, a5.shape)
a5
```
