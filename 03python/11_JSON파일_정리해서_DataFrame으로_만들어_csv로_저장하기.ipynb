* setdefault() 이용

```python
result2 = {}
for item in bank_data:
    for key, value in item.items():
        result2.setdefault(key, []).append(value)
result2
```

* defaultdict 이용

```python
from collections import defaultdict
```

```python
dic_result = defaultdict(list)
dic_result
```

```python
for item in bank_data:
    for key, value in item.items():
        dic_result[key].append(value)
result3 = dict(dic_result)
result3
```

```python
len(result3["사용자"])
```

```python
df = pd.DataFrame(result3)
df
```

* 강사님 코드

```python
result = {}
for item in bank_data:
    for key, value in item.items():
        if key == "별점":
            value = int(value)
        elif key == "리뷰":
            value = value.replace("?", " ").replace(".", "").replace("ㅋ", "")
        elif key == "개발자 응답":
            if value != None:
                value = value.replace("?", " ").replace(".", "").replace("ㅋ", "")
#        value = value.replace("?", " ").replace(".", "").replace("ㅋ", "") if value != None else "None" ->위의 두 줄 합쳐서 이렇게도 씀
        result.setdefault(key, []).append(value)
df = pd.DataFrame(result)
df
```

```python
df.info()
```

```python
df.to_csv("./data/국민은행_앱리뷰.csv", index = False, encoding = "utf-8-sig")
```

```python
print(len(data[0]["header"]["columns"]))
data[0]["header"]["columns"]
len(data[0]["body"]["items"][0].keys())
```

```python
result3 = {}
for page in data:
    for shop in page["body"]["items"]:
        for key, value in zip(data[0]["header"]["columns"], shop.values()):
            result3.setdefault(key, []).append(value)
df3 = pd.DataFrame(result3)
df3
```
