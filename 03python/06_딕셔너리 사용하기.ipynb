# 딕셔너리 사용하기
* 딕셔너리 dict타입, JSON형이라고도 함
* web에서 자료를 주고받을 때 주로 사용하는 방식(API)
* key: value 쌍으로 이루어진 자료형
* key를 호출하면 value가 나온다.
* 자료에 이름표를 붙인 형태
* key에는 list, dict를 제외한 자료형이 올 수 있다.
* value에는 모든 자료형의 데이터가 들어갈 수 있다.

# dict key의 자료형
* 문자열, 정수, 실수, 불린(True/False), tuple
* dict, list는 key가 될 수 없다.
* key는 변경되지 않는 값이어야 한다.

```python
d3 = {"백" : 100, 100 : "백", 3.5 : [2.5, 2.7], False : 0, (0, "서울") : "종로구"}
d3
```

```python
# list는 key로 올 수 없음
d4 = {"백" : 100, 100 : "백", 3.5 : [2.5, 2.7], False : 0, [0, "서울"] : "종로구"}
d4
```

```python
# dict는 key로 올 수 없음
d5 = {"백" : 100, 100 : "백", 3.5 : [2.5, 2.7], False : 0, {0 : "서울"} : "종로구"}
d5
```

```python
d3
```

```python
d3["백"]
```

```python
d3[100]
```

```python
d3[3.5]
```

```python
d3[False]
```

```python
d3[(0, "서울")]
```

# 딕셔너리 value의 자료형
* value에는 모든 자료형이 올 수 있다.

```python
d4 = {"정수" : 100, "실수" : 3.5, "문자열" : "딕셔너리의 값", 
      "리스트" : [1,2,3,4], "튜플" : (1,2,3,4), "딕셔너리" : {"key" : "value"}, "불린" : True, "set" : {1,2,3}}
d4
```

```python
d4["정수"]
```

```python
d4["set"]
```

```python
d4["리스트"]
```

```python
print(d4["리스트"][2])
```

```python
print(d4["딕셔너리"]["key"])
```

```python
print(d4["딕셔너리"]["key"], type(d4["딕셔너리"]))
```

# 딕셔너리에 자료 추가하기
* 변수["새로운 key"] = 자료
* 이미 딕셔너리에 같은 이름의 key가 있는 경우 덮어쓰기가 되어서 기존 자료가 사라짐

```python
a = 10
a = 50
print(a)
```

```python
d5 = {}
print(d5, type(d5))
```

```python
d5["번호"] = 0
d5
```

```python
d5[100] = "백"
d5
```

```python
d5["리스트"] = [1,2,3]
d5
```

* 이미 있는 key에 다른 자료 넣기

```python
d5["번호"] = 5
d5
```
